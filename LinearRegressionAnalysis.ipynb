```r
# Load necessary libraries
library(readr)

# Load the dataset
# Assuming the dataset is in CSV format and named 'data.csv'
data <- read_csv('data.csv')

# Display the first few rows of the dataset
head(data)
```

```r
# Fit a linear regression model
# Assuming 'y' is the dependent variable and 'x' is the independent variable
model <- lm(y ~ x, data = data)

# Display the summary of the model
summary(model)
```

```r
# Filter the dataset for the year 2016
# Assuming there is a 'year' column in the dataset
filtered_data <- data[data$year == 2016, ]

# Display the first few rows of the filtered dataset
head(filtered_data)
```

```r
# Fit a linear regression model using the filtered data
# Assuming 'y' is the dependent variable and 'x' is the independent variable
model_2016 <- lm(y ~ x, data = filtered_data)

# Display the summary of the model for 2016 data
summary(model_2016)
```
